#### Project Overview

This project analyses the result of two phase 3 randomized controlled trials - **EAGLE-2** and **EAGLE-3** -that compared a new oral antibiotic (**gepotidacin**) with the current standard treatment (**nitrofurantoin**) in women with uncomplicated unrinary tract infections (UTI)

The main goals are to:
1. Compare the therapeutic success rates of gepotidacin and nitrofurantoin
2. Testif there is a statistically significant difference in the estmated success rates
3. Test whether gepotidacin is non-inferior and superior to nitrofurantion (using a 10% non-inferiority margin and zero mark) respectively
4. Examine the difference between the results in the two trials.

This analysis uses two proportion tests and confidence intervals.

**Data sources:** Published results from theEAGLE-2 (NCT04020341) and EAGLE-3 (NCT04187144) trials.

```python
# Importing libraries

import pandas as pd
import numpy as np
from scipy import stats
import statsmodels.api  as sm
```

```python
# Hard coding the numbers from the trials result report

data = {
    'trials': ['EAGLE-2', 'EAGLE-2', 'EAGLE-3', 'EAGLE-3'],
    'treatment': ['gepotidacin', 'nitrofurantoin', 'gepotidacin', 'nitrofurantoin'],
    'successess': [162, 135, 162, 115],
    'total': [ 320, 287, 277, 264] 
}

df = pd.DataFrame(data)

df['failures'] = df['total'] - df['successess']

df
```

```python
# Creating df for EAGLE-2 

eagle_2 = df[df['trials'] == 'EAGLE-2']



print(eagle_2)
```

```python
# Calculating the success rates for both treatments in EAGLE-2

e2_ge = eagle_2[eagle_2['treatment'] == 'gepotidacin']
e2g_total = e2_ge['total'].iloc[0]
e2g_s_rate = (e2_ge['successess'] / e2_ge['total']).iloc[0]

e2_n = eagle_2[eagle_2['treatment'] == 'nitrofurantoin']
e2n_total = e2_n['total'].iloc[0]
e2n_s_rate = (e2_n['successess'] / e2_n['total']).iloc[0]


print(f'EAGLE-2 Gepotidacin success rate: {e2g_s_rate:.2%}')
print(f'EAGLE-2 nitrofurantoin success rate: {e2n_s_rate:.2%}')
```

```python
# Calculating the difference in success rate between gepotidacin and nitrofurantoin EAGLE-2

diff = e2g_s_rate - e2n_s_rate
diff = diff * 100

diff
```

Based on the success rate result:

Gepotidacin: 50.62%

Nitrofurantoin: 47.04%

Percentage point (in favour of gepotidacin): 3.58%


--

We can say that Gepotidacin has had a better effect on treating the uncomplicated urinary tract infections in women than nitrofurantoin, in the EAGLE-2. This estimate is unadjusted for other considerations of the patients in the trial.

```python
# calculating the unpooled standard error

se = np.sqrt(e2g_s_rate * (1 - e2g_s_rate) / e2g_total + e2n_s_rate * (1 - e2n_s_rate) / e2n_total)

se = se * 100

se
```

```python
# calculating confience interval. 95% confidence level is used 

lower_ci, upper_ci = stats.norm.interval(0.95, diff, se ) 

print(f'E2 Lower confidence interval: {lower_ci}')
print(f'E2 Upper confidence interval: {upper_ci}')
```

```python
# using a two sample z-test to verify if the difference in both trials are statistically significant in both trials. 
# Using the standard significance level 5%
# EAGLE-2

count_2 = [162, 135] # gipotidacin, nitrofurantoin
nob_2 = [320, 287]

z_stats_2, pvalue_2 = sm.stats.proportions_ztest(count_2, nob_2, alternative= 'two-sided')

print(f'z-stats: {z_stats_2:.2%}')
print(f'p-value: {pvalue_2:.2%}')
```

The confidence interval of the estimated difference in success rate between Gepotidacin and Nitrofurantoin has a lower and upper intervals of (-4.4% and 11.5%) respectively. This tells us that gepotidacin is non-inferior to nitrofurantoin

To determine if geptidacin is non-inferior or not on-inferior to nitrofurantoin, use a margin before the trial, in the EAGLE trials case, -10%:

- if the confidence interval is higher than the set margin of -10%, then gepotidacin  non-inferior to nitrofurantoin

To determine if geptidacin is superior or not superior to nitrofurantoin, use a margin of Zero (0):

- if the whole confidence interval is above zero, then gepotidacin is superior to nitrofurantoin.

To determine if our estmated success rate is statistically significant, we ran a hypothesis test
- **Null Hypothesis:** There is no difference in the success rate of treatments between gepotidacin and nitrofurantoin
- **Alternative Hypothesis:** There is a statistically significant difference in success rate between gepotidacin and nitrofurantoin

If the p-value from the two sample proportion z-tests is lower than our chose threshold (5% signifance level), then we reject the null hypothesis,
if it is higher than the significance level,  we fail to reject the null hypothesis. 

For EAGLE-2: p-value = 37.75%, p > 5%

In EAGLE-2 trial, our findings are:

1. Though gepotidacin based on this sample, showed a better effect of 3.58% percentage points, in treating uncomplicated urinary tract infections in women than nitrofurantoin, when checked for statistical significance using the two sample proportionsz-test, our p-value was abov or threshold level, therefore we fail to reject the null hypothesis. meaning, we have no statistical proof to support thata gepotidacin has a better treatment effect of UTI than nitrofurantoin.

2. Gepotidacin is non-inferior to nitrofurantoin, seeing that the lower confidence interval is above our threshold of -10%. 
3. Gepotidacin is not superior to nitrofurantoin, seeing that thewhole confidence interval is not above zero 

```python
# Creating a df for EAGLE-3

eagle_3 = df[df['trials'] == 'EAGLE-3']

print(eagle_3)
```

```python
# Calculating the success rates for both treatments in EAGLE-3 trial

e3_ge = eagle_3[eagle_3['treatment'] == 'gepotidacin']
e3g_total = e3_ge['total'].iloc[0]
e3g_s_rate = (e3_ge['successess'] / e3_ge['total']).iloc[0]

e3_n = eagle_3[eagle_3['treatment'] == 'nitrofurantoin']
e3n_total = e3_n['total'].iloc[0]
e3n_s_rate = (e3_n['successess'] / e3_n['total']).iloc[0]


print(f'EAGLE-3 Gepotidacin success rate: {e3g_s_rate:.2%}')
print(f'EAGLE-3 nitrofurantoin success rate: {e3n_s_rate:.2%}')
```

```python
# Calculate the difference in success rate between gepotidacin and nitrofurantoin EAGLE-3

diff3 = e3g_s_rate - e3n_s_rate
diff3 = diff3 * 100

diff3
```

Nitrofurantoin: 43.56%

```python
lower_ci, upper_ci = stats.norm.interval(0.95, diff3, se3 ) 

print(f'E3 Lower confidence interval: {lower_ci}')
print(f'E3 Upper confidence interval: {upper_ci}')
```

```python
# using a two sample z-test to verify if the difference in both trials are statistically significant in both trials. 
# Using the standard significance level 5%
# EAGLE-3


count_3 = [162, 115] # gipotidacin, nitrofurantoin
nob_3 = [277, 264]

z_stats_3, pvalue_3 = sm.stats.proportions_ztest(count_3, nob_3, alternative= 'two-sided')

print(f'z-stats: {z_stats_3:.2%}')
print(f'p-value: {pvalue_3:.2%}')
```

In the EAGLE-3 trial, the confidence interval of the estimated difference in success rate between Gepotidacin and Nitrofurantoin has a lower and upper intervals of (6.59% and 23.3%) respectively. This tells us that gepotidacin is non-inferior to nitrofurantoin in EAGLE-3 trial.

If the p-value from the two sample proportion z-tests is lower than our chose threshold (5% signifance level), then we reject the null hypothesis,
